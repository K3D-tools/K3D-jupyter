# Volume rendering: heart CT

The MetaImage heart scan shipped with the documentation assets, rendered as a
volume - first with the classic renderer, then under image-based lighting where
the volume reads its light from the environment map and casts ambient occlusion.

```python
import os

import numpy as np
import SimpleITK as sitk

import k3d

filepath = os.path.join(
    os.path.abspath(''),
    '..', 'docs', 'source', 'reference', 'assets', 'factory', 'heart.mhd')

img = sitk.GetArrayFromImage(sitk.ReadImage(filepath))

plot = k3d.plot()
plot += k3d.volume(img.astype(np.float32), samples=512,
                   alpha_coef=250, color_range=[0, 750])
plot.display()
```

## The same data under image-based lighting

`burnt_warehouse` carries one strong directional light - vessels shadow the heart
wall and the sulci read as depth thanks to the volume's occluder shell.

```python
plot.renderer = 'advanced'
plot.environment = 'burnt_warehouse'
```

```python
plot.tone_mapping = 'agx'
```
