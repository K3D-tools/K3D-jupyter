```python
import k3d
import ipywidgets as widgets
import numpy as np
```

```python
np.random.seed(2022)

x = np.random.randn(100, 3).astype(np.float32)

# Ten keyframes, keyed by time as a string.
frames = {str(t / 10): x - t / 5 * x / np.linalg.norm(x, axis=-1)[:, np.newaxis]
          for t in range(10)}
```

```python
plt_points = k3d.points(frames, color=0x528881, point_size=0.2)

plot = k3d.plot()
plot += plt_points
plot.display()
```

```python
# Which keyframes exist, without asking the browser.
plot.get_time_series_times()
```

```python
# [min, max] of those times; always contains 0.0, which is what time clamps to.
plot.get_time_series_range()
```

```python
# Discrete timesteps: hold each keyframe instead of blending towards the next one.
plot.time_interpolation = False
```

```python
# Steps from the keyframe nearest the current time and returns the new value.
plot.next_frame()
```

```python
plot.previous_frame()
```

```python
# Whole numbers of frames at a time; clamped at both ends, it does not wrap.
plot.step_frame(3)
```

```python
# Stepping is API only; wiring it to your own widgets takes no more than this:

previous = widgets.Button(description='Previous frame', icon='backward')
next_ = widgets.Button(description='Next frame', icon='forward')
readout = widgets.Label()


def show(_=None):
    readout.value = 'time = %g' % plot.time


previous.on_click(lambda _: (plot.previous_frame(), show()))
next_.on_click(lambda _: (plot.next_frame(), show()))
show()

widgets.HBox([previous, next_, readout])
```

```python
# Interpolation back on: time between keyframes blends them again.
plot.time_interpolation = True
plot.time = 0.45
```
